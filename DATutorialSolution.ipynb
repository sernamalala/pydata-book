### Question 2 - Using supplied data...

```python
url = 'https://raw.githubusercontent.com/adylw/Tutorials/refs/heads/main/tutorialDataA.csv'
pd.read_csv(url).head() #import data via URL
```

```python
data2 = pd.read_csv(url, skiprows=[0,1,3]) #import data via URL, do not read in any rows with text
data2
```

```python
data2.isna() #check rows or columns with only NA values (opt)
```

```python
data3 = data2.dropna() #remove remove rows or columns with only NA values
data3
```

```python
data4 = data2.fillna(data2['COST'].mean()) #replace/fill NA in cost column with mean of column data (in this case we must specify the cost column)
data4
```

_There a few ways to replace negative values within the dataframe. One way is to find the locations of negative values first and then use that list to index the replacement operation._

```python
data5 = data4.copy() #make a copy to work with
neg = [data5['COST'] < 0] #find negatives
neg
```

```python
data5['COST'].loc[neg[neg==True]]=42.069 #replace negative values with 42.069 using neg as an indexer
data5
```

_The other way is to find the locations of negative values and replace them in a single operation._

```python
data4['COST'][data4['COST'] < 0] = 42.069 #replace negative values with 42.069
data4
```

```python
data4 = data4.drop_duplicates() #remove duplicates (there was a mistake in the original data, my apologies)
data4
```

```python
data4 = data4.rename(columns=str.title) #change the column headings to title case
data4
```

```python
data4 = data4.sort_values('Year') #sort all data by year
data4
```

```python
pattern = r"(-[A-Z]{6})" #create the expression we seek to manipulate
regex = re.compile(pattern, flags=re.IGNORECASE)
data4['Season'] = data4['Season'].str.replace(regex,'-zone')
data4
#use regular expressions to read the rows of the seasons column and replace any matches of "-season name" with "-time".
```

```python
costChange=data4['Cost'].pct_change()*100 #calculate the percentage change of the cost column 
data4['Pct']=costChange.fillna(0) #add to dataframe, replace NA with 0.0 (opt)
data4
```

```python
data4.iloc[data4['Pct'].argmax()] #find the max change in the new column
```

```python
data4['Pct'].mean() #find the mean change in the new column
```

```python
#plot the percentage change and the cost on two separate graphs but in one figure (use subplot)
fig = plt.figure()
ax1=fig.add_subplot(1,2,1) 
ax2=fig.add_subplot(1,2,2)
#name the axes and title each graph
ax1.set(title="Cost per annum")
ax1.plot(data4["Year"],data4["Cost"],'-*')
ax1.set_xlabel("Year")
ax1.set_ylabel("Cost [Rands (ZAR)]")
ax2.set(title="Change in cost per annum")
ax2.plot(data4["Year"],data4["Pct"],'-*')
ax2.set_xlabel("Year")
ax2.set_ylabel("Change [%]")
#save the figure (in code) as JPG with a dpi of 600, paste the figure in your report(or print the jupyter notebook)
fig.savefig("tutorialDataB.jpg", dpi=600)
```

```python
data4.to_csv('tutorialDataB.csv', sep = ",")
```
