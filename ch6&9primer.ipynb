## 2. Basic writing

```python
arr2 = np.random.standard_normal(size=(100, 10)) #generate random values
```

```python
np.savetxt("new1.csv", arr2, delimiter=",") #write them to a file
```

```python
arr3 = pd.DataFrame(arr2) #there is a catch here... see plotting
arr3.to_csv("new2.csv")
```

```python
arr5 = np.loadtxt("new1.csv", delimiter=",") #read the same files
```

```python
arr6 = pd.read_csv("new2.csv",header=None)
```

## 3. Basic plotting

```python
plt.plot(arr5[1,:],color="black", linestyle="dotted", label="CT")
plt.plot(arr5[2,:],color="blue", linestyle="dashed", label="Jhb")
plt.plot(arr5[3,:],color="red", linestyle="solid", label="PE")
plt.plot(arr5[4,:],color="yellow", linestyle="dashdot", label="Pta")
plt.legend()
plt.xlabel("Stages")
plt.ylabel("Monthly frequency")
plt.title("Loadshedding Lottery")
plt.annotate("Thanks Eskom", xy=(8,1), xytext=(6,2), arrowprops=dict(facecolor="black", headwidth=10, width=2,
             headlength=6), horizontalalignment="left", verticalalignment="top")
plt.savefig("fig1.jpg", dpi=600)
```

```python
plt.plot(arr6.iloc[2][1:],color="black", linestyle="dotted", label="CT")
plt.plot(arr6.iloc[3][1:],color="blue", linestyle="dashed", label="Jhb")
plt.plot(arr6.iloc[4][1:],color="red", linestyle="solid", label="PE")
plt.plot(arr6.iloc[5][1:],color="yellow", linestyle="dashdot", label="Pta")
plt.legend()
plt.xlabel("Stages")
plt.ylabel("Monthly frequency")
plt.title("Loadshedding Lottery")
plt.annotate("Thanks Eskom", xy=(8,1), xytext=(6,2), arrowprops=dict(facecolor="black", headwidth=10, width=2,
             headlength=6), horizontalalignment="left", verticalalignment="top")
plt.savefig("fig2.jpg", dpi=600)
```
