### Exercise 3.2 - Extended

5 years = 365*5

```python
one_year = 365
# 0, 182, 365
specific_days = []
# use extend to add each element individually
for year in range(5):
    start = year*one_year
    specific_days.extend([start,start+182,start+365])

specific_data = []
for i in specific_days:
    day = new_file[new_file["Day"] == i]
    specific_data.append(day)

specific_data_df = pd.concat(specific_data,ignore_index=True)
```

```python
plt.plot(specific_data_df["Day"],specific_data_df["SD1neg"],linestyle="solid",label="SD1neg", color="green")
plt.plot(specific_data_df["Day"],specific_data_df["SD0"],linestyle="solid",label="SD0")
plt.plot(specific_data_df["Day"],specific_data_df["SD1"],linestyle="solid",label="SD1", color="yellow")

plt.xticks(specific_days)
plt.xlabel("Day")
plt.ylabel("SD1neg, SD0, and SD1")
plt.title("Day VS SD1neg, SD0, and SD1")
plt.legend()
```
